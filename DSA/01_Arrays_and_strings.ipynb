```python
def get_second_largest(nums):
    
    if len(nums) < 2:
        return None

    largest_num = float('-inf')
    second_num = float('-inf')

    for num in nums:
        if num > largest_num:
            second_num = largest_num
            largest_num = num

        elif largest_num > num > second_num:
            second_num = num

    if second_num == float('-inf'):
        return None   # no second largest

    return largest_num, second_num
```

```python

#    -> Rotate the list to the right by k positions.
#       Example: [1,2,3,4,5], k=2 → [4,5,1,2,3]
def rotate_list_v1(nums, k):
    return nums[-k:] + nums[:-k]

def rotate_list_v2(nums, k):
    for i in range(k):
        nums.insert(0, nums.pop())
    return nums

def rotate_list_v3(nums, k):
    while k > 0:
        nums.insert(0, nums.pop())
        k -= 1
    return nums

print(rotate_list_v3([1,2,3,4,5], 2))

```

```python
def merge_two_sorted_list_v1(list_1, list_2):
    list_1.extend(list_2)
    list_1.sort()
    return list_1

def merge_two_sorted_list_v2(list_1, list_2):
    new_list = []
    i = 0
    j = 0
    total_length = len(list_1) + len(list_2)
    for _ in range(total_length):
        if i >= len(list_1):
            new_list.append(list_2[j])
            j+=1
        elif j >= len(list_2):
            new_list.append(list_1[i])
            i += 1

        elif list_1[i] <= list_2[j]:
            new_list.append(list_1[i])
            i += 1
        else:
            new_list.append(list_2[j])
            j += 1

    return new_list


print(merge_two_sorted_list_v2([1, 3, 5], [2, 4, 6]))
```

```python
def sum_all(nums: List[int]) -> int:
    total = 0
    for i in range(len(nums)):
        total += nums[i]
    return total
```

```python
def find_first(nums: List[int], x: int) -> int:
    for index, num in enumerate(nums):
        if num == x:
            return index
    return -1
```

```python
print("find_first(20):", find_first(nums1, 20))
print("find_first(7):", find_first(nums1, 7))
```

```python
def two_sum_sorted(nums: List[int], target: int) -> Optional[Tuple[int, int]]:
    i, j = 0, len(nums) -1
    while i < j:
        s = nums[i] + nums[j]
        if s == target:
            return i, j
        elif s < target:
            i += 1
        else:
            j -= 1
    return None 

list_1 = [5, 10, 20, 40, 50, 60, 70]
print("\ntwo_sum_sorted(list_1, 90):", two_sum_sorted(list_1, 90))
```

```python
def move_zeros_to_end(nums: List[int]) -> List[int]:
    left, right = 0, len(nums)-1
    while left < right:
        if nums[left] != 0:
            left += 1
        elif nums[right] == 0:
            right -= 1
        else:
            nums[left], nums[right] = nums[right], nums[left]
            left += 1
            right -= 1
    return nums
```

```python
arr = [0, 1, 0, 3, 12]
print("move_zeros_to_end", arr[:], move_zeros_to_end(arr[:])) 
```

```python
strs = ["flower", "flow", "flight"]

# find shortest word
words_lens = {}
for i in range(len(strs)):
    words_lens[strs[i]] = len(strs[i])

choose_min = min(words_lens.values())
choose_min_word = [word for word in strs if len(word) == choose_min]

# build prefixes from shortest word
search = []
for i in range(len(choose_min_word[0])):
    search.append(choose_min_word[0][:i+1])

# check prefixes
result = ""
for prefix in search:
    for word in strs:
        if prefix not in word:
            break
    else:
        result = prefix

print(result)
```

```python
strs = ["flower", "flow", "flight"]

shortest = min(strs, key=len)
print(shortest)
result = ""
for i in range(len(shortest)):
    prefix = shortest[:i+1]
    if all(word.startswith(prefix) for word in strs):
        result = prefix
    else:
        break

print(result)
```

```python
list_1 = [0, 3, 2, 5, 1, 9, 7, 10]

def sort_list(list_1):
    n = len(list_1)

    for pass_num in range(n):
        print(f"\n🔁 PASS {pass_num + 1}")
        for i in range(n - 1):
            print(f"  Compare {list_1[i]} and {list_1[i+1]} → ", end="")

            if list_1[i] > list_1[i + 1]:
                list_1[i], list_1[i + 1] = list_1[i + 1], list_1[i]
                print(f"swap → {list_1}")
            else:
                print("no swap")

        print(f"End of pass {pass_num + 1}: {list_1}")

    return list_1


print("\n✅ Final Sorted List:")
print(sort_list(list_1))
```
