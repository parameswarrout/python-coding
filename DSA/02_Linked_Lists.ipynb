```python
class LinkedList:
    def __init__(self):
        self.head = None

    # ---------------- UTILITIES ---------------- #

    def is_empty(self):
        return self.head is None

    def length(self):
        count = 0
        temp = self.head
        while temp:
            count += 1
            temp = temp.next
        return count

    def search(self, data):
        index = 0
        temp = self.head
        while temp:
            if temp.data == data:
                return index
            temp = temp.next
            index += 1
        return -1

    # ---------------- INSERT ---------------- #

    def insert_at_beginning(self, data):
        new_node = Node(data)
        new_node.next = self.head
        self.head = new_node

    def insert_at_end(self, data):
        new_node = Node(data)

        if self.head is None:
            self.head = new_node
            return

        temp = self.head
        while temp.next:
            temp = temp.next

        temp.next = new_node

    def insert_at_position(self, data, pos):
        if pos < 0:
            raise IndexError("Position must be non-negative")

        new_node = Node(data)

        # insert at head
        if pos == 0:
            new_node.next = self.head
            self.head = new_node
            return

        if self.head is None:
            raise IndexError("Position out of range")

        temp = self.head
        for _ in range(pos - 1):
            if temp.next is None:
                raise IndexError("Position out of range")
            temp = temp.next

        new_node.next = temp.next
        temp.next = new_node

    # ---------------- DELETE ---------------- #

    def delete_from_beginning(self):
        if self.head is None:
            return
        self.head = self.head.next

    def delete_from_end(self):
        if self.head is None:
            return

        if self.head.next is None:
            self.head = None
            return

        temp = self.head
        while temp.next.next:
            temp = temp.next

        temp.next = None

    def delete_at_position(self, pos):
        if pos < 0:
            raise IndexError("Position must be non-negative")

        if self.head is None:
            raise IndexError("Position out of range")

        if pos == 0:
            self.head = self.head.next
            return

        temp = self.head
        for _ in range(pos - 1):
            if temp.next is None:
                raise IndexError("Position out of range")
            temp = temp.next

        if temp.next is None:
            raise IndexError("Position out of range")

        temp.next = temp.next.next

    # ---------------- REVERSE LINK LIST ---------------- #  
    def reverse_link_list(self):
        prev = None
        current = self.head
        
        while current:
            next_node = current.next  # 1. Save the next node (so we don't lose it)
            current.next = prev       # 2. Reverse the pointer (point backward)
            prev = current            # 3. Move prev forward
            current = next_node       # 4. Move current forward
            
        self.head = prev  # 5. Update head to the new front of the list
         

    # ---------------- PRINT ---------------- #

    def print_list(self):
        temp = self.head
        while temp:
            print(temp.data, end=" → ")
            temp = temp.next
        print("None")
```

```python
ll = LinkedList()

# Insert at beginning
ll.insert_at_beginning(30)
ll.insert_at_beginning(20)
ll.insert_at_beginning(10)
ll.print_list()
# 10 → 20 → 30 → None

# Insert at end
ll.insert_at_end(40)
ll.insert_at_end(50)
ll.print_list()
# 10 → 20 → 30 → 40 → 50 → None

# Insert at position
ll.insert_at_position(25, 2)
ll.print_list()
# 10 → 20 → 25 → 30 → 40 → 50 → None

# Delete from beginning
ll.delete_from_beginning()
ll.print_list()
# 20 → 25 → 30 → 40 → 50 → None

# Delete at position
ll.delete_at_position(2)
ll.print_list()
# 20 → 25 → 40 → 50 → None

# Delete from end
ll.delete_from_end()
ll.print_list()
# 20 → 25 → 40 → None
```

```python
# ---------------- EXTRA TESTS ---------------- #
ll = LinkedList()
assert ll.is_empty() is True
assert ll.length() == 0

ll.insert_at_beginning(10)
ll.insert_at_end(20)
ll.insert_at_end(30)
assert ll.length() == 3
assert ll.search(20) == 1
assert ll.search(99) == -1

ll.insert_at_position(15, 1)
assert ll.search(15) == 1

ll.delete_at_position(1)
assert ll.search(15) == -1

try:
    ll.insert_at_position(5, -1)
except IndexError:
    print("insert negative position: OK")

try:
    ll.delete_at_position(10)
except IndexError:
    print("delete out of range: OK")

empty_ll = LinkedList()
try:
    empty_ll.delete_at_position(0)
except IndexError:
    print("delete on empty list: OK")
```

```python
class DoubleListNode:
    def __init__(self, val=0, prev=None, next=None):
        self.data = val
        self.prev = prev
        self.next = next


class DoublyLinkedList:
    
    def __init__(self):
        self.head = None
        self.tail = None
        self.size = 0
        
        
    # ---------------- UTILITIES ---------------- #
    def is_empty(self):
        return self.head is None
    
    def length(self):
        count = 0
        temp = self.head
        while temp:
            count += 1
            temp = temp.next
        return count
    
    def search(self, data):
        index = 0
        temp = self.head
        while temp:
            if temp.data == data:
                return index
            temp = temp.next
            index += 1
        return -1
    
    # ---------------- INSERT ---------------- #
    
    def insert_at_beginning(self, data):
        new_node = DoubleListNode(data)

        # Case 1: Empty list
        if self.head is None:
            self.head = new_node
            self.tail = new_node
        else:
            new_node.next = self.head
            new_node.prev = None
            self.head.prev = new_node
            self.head = new_node

        self.size += 1
        
    def insert_at_end(self, data):
        new_node = DoubleListNode(data)
        
        if self.head is None:
            self.head = new_node
            self.tail = new_node
            self.size += 1
            return 
            
        else:
            self.tail.next = new_node
            new_node.prev = self.tail
            self.tail = new_node
            
        self.size += 1
            
```
