# Prefix Sum - 100 Practice Questions

## What is Prefix Sum?
Prefix sum is a technique where we precompute cumulative sums to answer range sum queries efficiently.

### Basic Formula:
- `prefix[i] = arr[0] + arr[1] + ... + arr[i]`
- Range sum from index `left` to `right`: `prefix[right] - prefix[left-1]`

### Template Code:
```python
# Build prefix sum array
prefix = [0] * n
prefix[0] = arr[0]
for i in range(1, n):
    prefix[i] = prefix[i-1] + arr[i]

# Range sum query (left to right)
def range_sum(prefix, left, right):
    if left == 0:
        return prefix[right]
    return prefix[right] - prefix[left-1]
```
