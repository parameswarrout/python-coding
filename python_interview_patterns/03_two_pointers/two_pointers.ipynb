# Two Pointers Technique

This notebook covers the two pointers technique for solving array and string problems.

## What is the Two Pointers Technique?
The two pointers technique uses two integer variables (pointers) that iterate through a data structure (typically an array or string) concurrently. This approach is highly efficient for optimizing time complexity from $O(n^2)$ to $O(n)$ or $O(log n)$.

### Common Patterns:
1. **Opposite Ends (Left & Right):** One pointer starts at the beginning (index 0) and the other starts at the end (index $n-1$). They move toward each other until they meet. (Used for: Palindromes, sorted Two Sum, reversing arrays).
2. **Fast & Slow:** Both pointers start at the same position but move at different speeds. (Used for: Cycle detection in linked lists, finding middle of a list).
3. **Sliding Window Boundary:** Pointers represent the start and end of a subarray/substring. (Used for: Subarray sum, substring search).
