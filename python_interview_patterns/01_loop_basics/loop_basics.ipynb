```python
# 1. Print numbers from 1 to n
def print_1_to_n(n):
    for i in range(1, n + 1):
        print(i)


# 2. Print numbers from n to 1
def print_n_to_1(n):
    for i in range(n, 0, -1):
        print(i)


# 3. Print all even numbers from 0 to n
def print_even_numbers(n):
    for i in range(0, n + 1):
        if i % 2 == 0:
            print(i)


# 4. Print all odd numbers from 1 to n
def print_odd_numbers(n):
    for i in range(1, n + 1):
        if i % 2 != 0:
            print(i)


# 5. Print all multiples of k that are <= n
def print_multiples_of_k(n, k):
    i = k
    while i <= n:
        print(i)
        i += k


# 6. Count numbers divisible by 3 from 1 to n
def count_divisible_by_3(n):
    count = 0
    for i in range(1, n + 1):
        if i % 3 == 0:
            count += 1
    return count


# 7. Sum of numbers from 1 to n (using loop)
def sum_1_to_n(n):
    total = 0
    for i in range(1, n + 1):
        total += i
    return total


# 8. Sum of squares from 1^2 to n^2
def sum_of_squares(n):
    total = 0
    for i in range(1, n + 1):
        total += i ** 2
    return total


# 9. Factorial of n
def factorial(n):
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result


# 10. Count number of digits in n
def count_digits(n):
    if n == 0:
        return 1

    count = 0
    while n > 0:
        count += 1
        n //= 10
    return count
```

```python
nums = [7, 12, -5, 0, 23, 8, 17, 70, 19, -11, 4, 29, 35, 2, 97, 14, -7, 41, 6, 73]
k = 7      # for divisible by k
x = 10     # for greater than x
a, b = 5, 30   # for range [a, b]
n = 100    # for primes between 1 and n
```

```python
# 21
def Count_positive_negative_zeros(nums):
    positive, negative, zeros = 0, 0, 0
    for num in nums:
        if num > 0:
            positive += 1
        elif num < 0:
            negative += 1
        else:
            zeros += 1
    return positive, negative, zeros

# 22
def even_sum(nums):
    result = 0
    for num in nums:
        if num % 2 == 0:
            result += num
    return result

# 23
def sum_divisible_by_k(nums, k):
    result = 0
    for num in nums:
        if num % k == 0:
            result += num
    return result

# 24
def count_greater_than_x(nums, x):
    COUNT = 0
    for num in nums:
        if num > x:
            COUNT += 1
    return COUNT 

# 25
def count_in_range(nums, a, b):
    result = 0
    for num in nums:
        if num >= a and num <= b:
            result += 1
    return result

# 26
def find_min_max(nums):
    minimum = maximum = nums[0]
    for num in nums:
        if num > maximum:
            maximum = num
        if num < minimum:
            minimum = num
    return minimum, maximum

# 27
def second_largest(nums):
    largest = float('-inf')
    sec_largest = float('-inf')
    nums = list(set(nums))
    if len(nums) < 2:
        return None
    for num in nums:
        if num > largest:
            sec_largest = largest
            largest = num
        elif largest > num > sec_largest:
            sec_largest = num
    if sec_largest == float('-inf'):
        sec_largest = None
    return sec_largest
# 28
def count_primes_upto_n(n):
    def is_prime(x):
        if x < 2:
            return False
        for num in range(2, int(x ** 0.5)+1):
            if x % num == 0:
                return False
        return True
    COUNT = 0
    for i in range(2, n+1):
        if is_prime(i):
            COUNT += 1
    return COUNT

# 29
def count_primes_in_list(nums):
    def is_prime(n):
        if n < 2:
            return False
        for i in range(2, int(n ** 0.5)+1):
            if n % i == 0:
                return False
        return True
    COUNT = 0
    for num in nums:
        if is_prime(abs(num)):
            COUNT += 1
    return COUNT


# 30
def count_numbers_with_digit_7(nums):
    COUNT = 0
    for num in nums:
        if '7' in str(num):
            COUNT += 1
    return COUNT

print("21. Count_positive_negative_zeros:", Count_positive_negative_zeros(nums))
print("22. even_sum:", even_sum(nums))
print("23. sum_divisible_by_k:", sum_divisible_by_k(nums, k))
print("24. count_greater_than_x:", count_greater_than_x(nums, x))
print("25. count_in_range:", count_in_range(nums, a, b))
print("26. find_min_max:", find_min_max(nums))
print("27. second_largest:", second_largest(nums))
print("28. count_primes_upto_n:", count_primes_upto_n(n))
print("29. count_primes_in_list:", count_primes_in_list(nums))
print("30. count_numbers_with_digit_7:", count_numbers_with_digit_7(nums))
```

## Interview Practice - Set 4 (31-40)

31. Count vowels in a string.
32. Count consonants in a string.
33. Reverse a string using a loop.
34. Build a new string of every second character (index 0,2,4,...).
35. Check if a string is a palindrome using a loop.
36. Count runs of consecutive equal characters (e.g., "aabcc" has runs aa, b, cc).
37. Run-length encode a string (e.g., "aabcc" -> "a2b1c2").
38. Toggle case for each character in a string.
39. Remove all digits from a string.
40. Count how many times a 2-char substring appears (overlapping allowed).


```python
def count_vowels(s):
    COUNT = 0
    for i in s:
        if i.lower() in "aeiou":
            COUNT += 1
    return COUNT

def count_consonants(s):
    COUNT = 0
    for i in s:
        if i.lower() not in "aeiou":
            COUNT += 1
    return COUNT

def reverse_string(s):
    s = list(s)
    start = 0
    end = len(s)-1
    while start < end:
        s[start], s[end] = s[end], s[start]
        start += 1
        end -= 1
    return "".join(s)

def reverse_string_1(s):
    result=""
    for i in range(len(s)-1, -1, -1):
        result += s[i]
    return result
        
def every_second_char(s):
    result = ""
    for i in range(0, len(s), 2):
        result += s[i]
    return result

def is_palindrome(s):
    start = 0
    end = len(s)-1
    while start < end:
        if s[start] != s[end]:
            return False
        start +=1
        end -= 1
    return True

def count_runs(s):
    run = 1
    if len(s) < 1:
        return 0
    for i in range(1,len(s)):
        if s[i] != s[i-1]:
            run += 1
    return run

def run_length_encode(s):
    if len(s) == 0:
        return ""

    result = ""
    run = 1

    for i in range(1, len(s)):
        if s[i] == s[i-1]:
            run += 1
        else:
            result += s[i-1] + str(run)
            run = 1

    result += s[-1] + str(run)

    return result


def toggle_case(s):
    result = ""

    for ch in s:
        if ch.islower():
            result += ch.upper()
        elif ch.isupper():
            result += ch.lower()
        else:
            result += ch

    return result


def remove_digits(s):
    result = ""
    for ch in s:
        if not ch.isdigit():
            result += ch

    return result

def count_substring_2(s, sub):
    if len(s) == 0 or len(s) < len(sub):
        return 0
    count = 0
    for i in range(len(s) - len(sub)+1):
        if s[i:i+len(sub)] == sub:
            count += 1
    return count


print(count_vowels("Hello World"))
print(count_consonants("Hello World"))
print(reverse_string_1("Python"))
print(every_second_char("abcdefg"))
print(is_palindrome("madam"))
print(is_palindrome("hello"))
print(count_runs("aabcc"))
print(run_length_encode("aabcc"))
print(toggle_case("PyThOn"))   # pYtHoN
print(toggle_case("Hello123")) # hELLO123
print(remove_digits("a1b2c3d4"))
print(count_substring_2("aaaa", "aa"))
```

## Interview Practice - Set 5 (41-50)

41. Given nums, create a new list of squares.
42. Given nums, create a new list of cubes.
43. Filter nums to keep only non-negative values.
44. Remove duplicates from nums while preserving order (no set).
45. Merge two lists by alternating elements; append leftovers.
46. Split nums into two lists: evens and odds.
47. Return a list of all indices where nums[i] == target.
48. Move all zeros to the end of nums (stable order).
49. Remove all occurrences of target from nums using a loop (return new list).
50. Generate the first n Fibonacci numbers using a loop.


```python
def square_numbers(numbers):
    result = []
    for n in numbers:
        result.append(n ** 2)
    return result


def cube_numbers(numbers):
    result = []
    for n in numbers:
        result.append(n ** 3)
    return result


def filter_non_negative(numbers):
    result = []
    for n in numbers:
        if n >= 0:
            result.append(n)
    return result


def remove_duplicates(numbers):
    seen = set()
    result = []
    for n in numbers:
        if n not in seen:
            seen.add(n)
            result.append(n)
    return result


def merge_lists_while(list1, list2):
    result = []
    index = 0
    limit = min(len(list1), len(list2))

    while index < limit:
        result.append(list1[index])
        result.append(list2[index])
        index += 1

    result.extend(list1[index:] or list2[index:])
    return result


def merge_lists_for(list1, list2):
    result = []
    limit = min(len(list1), len(list2))

    for i in range(limit):
        result.append(list1[i])
        result.append(list2[i])

    result.extend(list1[limit:] or list2[limit:])
    return result


def split_even_odd(numbers):
    even = []
    odd = []

    for n in numbers:
        if n in (0, 1):
            continue
        if n % 2 == 0:
            even.append(n)
        else:
            odd.append(n)

    return even, odd


def find_indices(numbers, target):
    result = []
    for i, n in enumerate(numbers):
        if n == target:
            result.append(i)
    return result


def move_zeros_to_end(numbers):
    pos = 0
    for i in range(len(numbers)):
        if numbers[i] != 0:
            numbers[pos], numbers[i] = numbers[i], numbers[pos]
            pos += 1
    return numbers


def remove_target_value(numbers, target):
    result = []
    for n in numbers:
        if n != target:
            result.append(n)
    return result


def fibonacci_numbers(n):
    result = []
    first, second = 0, 1

    if n >= 1:
        result.append(first)
    if n >= 2:
        result.append(second)

    for _ in range(n - 2):
        first, second = second, first + second
        result.append(second)

    return result



print("Squares:", square_numbers([1, 2, 3, 4]))
print("Cubes:", cube_numbers([1, 2, 3]))
print("Non-negative:", filter_non_negative([-2, -1, 0, 3, 5]))
print("Remove duplicates:", remove_duplicates([1, 2, 2, 3, 1, 4]))

print("Merge (while):", merge_lists_while([1, 2, 3, 4], [10, 20, 30]))
print("Merge (for):", merge_lists_for([1, 2, 3, 4], [10, 20, 30]))

print("Even & Odd:", split_even_odd([0, 1, 2, 3, 4, 5, 6]))
print("Indices:", find_indices([1, 3, 7, 3, 9, 3], 3))
print("Move zeros:", move_zeros_to_end([0, 1, 0, 3, 12]))
print("Remove target:", remove_target_value([1, 2, 3, 2, 4], 2))
print("Fibonacci:", fibonacci_numbers(7))
```

## Interview Practice - Set 6 (51-60)

51. Print all pairs (i, j) for i in 1..n and j in 1..n.
52. Count pairs (i, j) with i < j and nums[i] == nums[j].
53. Count pairs with even sum in nums (i < j).
54. Print all index pairs (i, j) where nums[i] + nums[j] == target.
55. Return True if any pair sums to target, else False.
56. Count inversions: pairs (i, j) with i < j and nums[i] > nums[j].
57. Find common elements between two lists without using set.
58. Build an n x n multiplication table as a list of lists.
59. Print all 2-character combinations from a string (i < j).
60. Given list of strings, find all pairs that share the same first letter.


```python
# 51
def print_all_pairs_nxn(n):
    result = []
    for i in range(1, n + 1):
        for j in range(1, n + 1):
            result.append((i, j))
    return result


# 52
def count_equal_pairs(nums):
    count = 0
    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] == nums[j]:
                count += 1
    return count


# 53
def count_even_sum_pairs(nums):
    count = 0
    for i in range(len(nums)):
        for j in range(i+1, len(nums)):
            if nums[i] % 2 == nums[j] % 2:
                count += 1
    return count


# 54
def print_pairs_with_target_sum(nums, target):
    result = []
    for i in range(len(nums)):
        for j in range(i+1, len(nums)):
            if nums[i] + nums[j] == target:
                result.append((i, j))
    return result


def print_pairs_with_target_sum_v2(nums, target):
    result = []
    seen = {}
    for i in range(len(nums)):
        complement = target - nums[i]
        if complement in seen:
            result.append((seen[complement], i))
        seen[nums[i]] = i
    return result


# 55
def has_pair_with_target_sum(nums, target):
    seen = set()
    for num in nums:
        complement = target - num
        if complement in seen:
            return True
        seen.add(num)
    return False


# 56
def count_inversions(nums):
    count = 0
    for i in range(len(nums)):
        for j in range(i+1, len(nums)):
            if nums[i] > nums[j]:
                count += 1
    return count


# 57
def common_elements_without_set(list1, list2):
    result = []
    for num in list1:
        if num in list2:
            result.append(num)
    return result


# 58
def build_multiplication_table(n):
    result = []
    for i in range(1, n+1):
        row = []
        for j in range(1, n+1):
            row.append(i*j)
        result.append(row)
    return result


# 59
def print_two_char_combinations(s):
    result = []
    for i in range(len(s)):
        for j in range(i+1, len(s)):
            result.append(s[i] + s[j])
    return result


# 60
def pairs_with_same_first_letter(words):
    result = []
    for i in range(len(words)):
        for j in range(i+1, len(words)):
            if words[i][0] == words[j][0]:
                result.append((words[i], words[j]))
    return result

print("51:", print_all_pairs_nxn(3))

nums = [1, 2, 2, 3, 4]
print("52:", count_equal_pairs(nums))
print("53:", count_even_sum_pairs(nums))

print("54 brute:", print_pairs_with_target_sum(nums, 4))
print("54 optimized:", print_pairs_with_target_sum_v2(nums, 4))

print("55:", has_pair_with_target_sum(nums, 5))

print("56:", count_inversions([8, 4, 2, 1]))

print("57:", common_elements_without_set([1,2,3,4], [3,4,5,6]))

print("58 multiplication table:")
table = build_multiplication_table(4)
for row in table:
    print(*row)

print("59:", print_two_char_combinations("abcd"))

print("60:", pairs_with_same_first_letter(["apple", "ant", "banana", "ball", "cat"]))
```
