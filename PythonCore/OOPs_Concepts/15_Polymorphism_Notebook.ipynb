# Polymorphism in Object-Oriented Programming
## Definition
Polymorphism means "many forms" and refers to the ability of different objects to respond to the same method call in different ways. It allows a single interface to represent different underlying data types or classes.
## Key Concepts
- Same method name, different implementations
- Runtime decision-making about which method to execute
- Enables flexible and extensible code

## Types of Polymorphism

### 1. Method Overriding (Runtime Polymorphism)
Child classes override parent class methods with their own implementation.

### 2. Duck Typing
"If it walks like a duck and quacks like a duck, then it's a duck." Objects are defined by their behavior rather than their type.

```python
class Animal:
    
    def sound(self):
        pass
    
    
class Dog(Animal):
    
    def sound(self):
        return "Woof!"
    
    
class Cat(Animal):
    
    def sound(self):
        return "Meow!"
    
def make_animal_sound(animal):
    return animal.sound()
```

```python
dog = Dog()
cat = Cat()

print(make_animal_sound(dog))
print(make_animal_sound(cat))
```

```python
# len() works differently for different data types
print(len("hello"))           # Output: 5 (for string)
print(len([1, 2, 3, 4]))      # Output: 4 (for list)
print(len({"a": 1, "b": 2}))  # Output: 2 (for dictionary)
```

```python
class Point:
    
    def __init__(self, x, y):
        self.x = x
        self.y = y
        
    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)

    def __str__(self):
        return f"Point({self.x},  {self.y})"
```

```python
p1 = Point(1, 2)
p2 = Point(3, 4)
p3 = Point(5, 1)
p4 = p1 + p2 + p3 
print(p4)  
```

```python
from math import pi

class Shape:
    
    def area(self):
        pass
    
class Rectangle(Shape):
    
    def __init__(self, width, height):
        self.width = width
        self.height = height
        
    def area(self):
        return self.width * self.height
    
class Circle(Shape):
    
    def __init__(self, radius):
        self.radius = radius
        
    def area(self):
        return round(pi * self.radius ** 2, 2)
    
def print_area(shape):
    print(f"Area: {shape.area()}")
```

```python
shapes = [Rectangle(5, 3), Circle(4)]
for shape in shapes:
    print_area(shape)
```

abstractmethod = a rule that forces every child class to implement that method, or it cannot be used.

```python
from abc import ABC, abstractmethod

class Vehicle(ABC):
    @abstractmethod
    def start_engine(self):
        pass
    
class Car(Vehicle):
    def start_engine(self):
        return "Car engine started"

class Motorcycle(Vehicle):
    def start_engine(self):
        return "Motorcycle engine started"
```

```python
class A:
    def show(self):
        return "A"

class B:
    def show(self):
        return "B"

class C(A, B):
    pass

c = C()
print(c.show())  # Uses MRO (Method Resolution Order)
```
