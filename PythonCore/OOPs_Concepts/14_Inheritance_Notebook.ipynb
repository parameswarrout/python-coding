## Inheritance allows a class to inherit attributes and methods from another class.

```python
class Animal:
    
    def __init__(self, name, species):
        self.name = name
        self.species = species
        
    def make_sound(self):
        return f"{self.name} make a sound"
    
    def sleep(self):
        return f"{self.name} is sleeping."
    
    def info(self):
        return f"{self.name} is a {self.species}"
```

```python
generic_animal = Animal("Generic", "Unknown")
print("--- Generic Animal ---")
print(generic_animal.info())
print(generic_animal.make_sound())
print(generic_animal.sleep())
```

```python
class Dog(Animal):
    
    def __init__(self, name, species, breed):
        super().__init__(name, "Dog")
        self.breed = breed
        
    def make_sound(self):
        return f"{self.name} braks!"
    
    def fetch(self):
        return f"{self.name} fetches the ball!"
```

```python
Dog = Animal("Buddy", "Golden Retriever")
print("--- Animal ---")
print(Dog.info())
print(Dog.make_sound())
print(Dog.sleep())
```

```python
class Cat(Animal):
    
    def __init__(self, name, color):
        super().__init__(name, "cat")
        self.color = color 
        
        
    def make_sound(self): 
        return f"{self.name} meows!"

    def climb(self):  
        return f"{self.name} climbs the tree!"
        
```

```python
my_cat = Cat("Whiskers", "Orange")
print(f"Info: {my_cat.info()}")          # From parent class
print(f"Sound: {my_cat.make_sound()}")    # Overridden method
print(f"Sleep: {my_cat.sleep()}")         # From parent class
print(f"Climb: {my_cat.climb()}")         # Specific to Cat class
print()
```

```python
class Vehicle:
    def __init__(self, name, brand, speed, wheels):
        self.name = name 
        self.brand = brand
        self.speed = speed
        self.wheels = wheels
        
    def move(self):
        return "Vehicle is moving."
    
    def info(self):
        return f"Brand: {self.brand}, Speed: {self.speed}, Wheels: {self.wheels}"
```

```python
class Car(Vehicle):
    
    def __init__(self, name, fuel_type, brand, speed):
        super().__init__(name, brand, speed, 4)   
        self.fuel_type = fuel_type
        
    def start(self):
        return f"{self.name} has started."
    
    def honk(self):
        return f"{self.name} honked."
```

```python
Audi = Car("Q7", "Petrol", "Audi", 240)

print(Audi.start())
print(Audi.honk())
print(Audi.info())
```

```python
class Bike(Vehicle):
    
    def __init__(self, name, brand, speed):
        super().__init__(name, brand, speed, 2)
        
    def start_engine(self):
        return f"{self.name} has started."
```

```python
Bullet = Bike("Intercepter 650", "Royal Enfield", 250)

print(Bullet.start_engine())
print(Bullet.info())
```
