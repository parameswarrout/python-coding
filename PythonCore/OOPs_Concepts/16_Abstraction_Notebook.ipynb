# Abstraction
#### Abstraction means showing only what is necessary and hiding how it works.
#### Abstraction means defining what a class must do, not how it does it.

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod   # Every Shape must have area()
    def area(self): 
        pass

class Square(Shape):
    def area(self):
        return 4 * 4
    
s = Square()
print(s.area())

```

```python
# Without abstraction
class Paytm:
    
    def pay(self, amt):
        print("Paytm", amt)
        
class Gpay:
    def pay(self, amt):
        print("gpay", amt)
```

```python
# With abstraction
class Payment(ABC):
    @abstractmethod
    def pay(self, amt):
        pass
    
# Now every payment system must follow same rule.
```

```python
def process(p: Payment):
    p.pay(100)

process(Paytm())
process(Gpay())
```

```python
class Vehicle(ABC):
    def fuel_type(self):
        return "Petrol or Diesel"

    @abstractmethod
    def start(self):
        pass
```

```python
class Car(Vehicle):
    def start(self):
        return "Car started"
```

```python
class Account(ABC):
    @property
    @abstractmethod
    def balance(self):
        pass
```

```python
class Bank(Account):
    def __init__(self, bal):
        self._bal = bal

    @property
    def balance(self):
        return self._bal
```

```python
class Report(ABC):
    def generate(self):
        self.header()
        self.body()
        self.footer()

    def header(self):
        print("Company Name")

    @abstractmethod
    def body(self):
        pass

    def footer(self):
        print("Thank you")
```

```python
class SalesReport(Report):
    def body(self):
        print("Sales Data")
```

```python
class Device(ABC):
    @abstractmethod
    def on(self): pass

    @abstractmethod
    def off(self): pass
```

```python
class Flyable(ABC):
    @abstractmethod
    def fly(self): pass
```
